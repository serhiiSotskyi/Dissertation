```python
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd().resolve().parents[0]
summary_df = pd.read_csv(PROJECT_ROOT / 'outputs_v2' / 'metrics' / 'synthetic_fusion_summary.csv')
summary_df
```

```python
pivot = summary_df.pivot_table(index=['pairing_type', 'experiment'], columns='metric', values='mean')
pivot
```

## Interpretation
The same-label synthetic pairing is expected to overstate alignment because labels are used during construction. The random-pairing control is therefore essential. These results are not evidence of clinical multimodal benefit; they are evidence about the behavior of fusion methods under synthetic pairing.
