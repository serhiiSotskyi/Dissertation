# BreaKHis Patient-Level Dataset

## Objective
Create a defendable BreaKHis split at patient level and prove that patient leakage is removed.

## Why this matters
The legacy image notebooks used image-level random splitting, which can place images from the same patient into multiple subsets. For medical imaging, that inflates performance and is not defendable in the dissertation.

```python
import sys
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd().resolve().parents[0]
sys.path.append(str(PROJECT_ROOT))

from src.breakhis import build_binary_records, cohort_summary, patient_level_split, patient_overlap_report, split_summary, sample_per_patient

dataset_root = PROJECT_ROOT / 'data' / 'dataset_cancer_v1' / 'dataset_cancer_v1'
df = build_binary_records(dataset_root)
print('Total images:', len(df))
print('Total patients:', df['patient_id'].nunique())
df.head()
```

```python
cohort_summary(df)
```

```python
split = patient_level_split(df, random_state=42)
summary = split_summary(split)
overlap = patient_overlap_report(split)
print(summary.to_string(index=False))
print(overlap)
```

```python
sampled_train = sample_per_patient(split.train, per_magnification=1, random_state=42)
sampled_val = sample_per_patient(split.val, per_magnification=1, random_state=42)
sampled_test = sample_per_patient(split.test, per_magnification=1, random_state=42)
print('Sampled train rows:', len(sampled_train))
print('Sampled val rows:', len(sampled_val))
print('Sampled test rows:', len(sampled_test))
```

## Interpretation
The overlap dictionary must be all zeros. The sampled subsets are used later for the lightweight patient-level embedding baseline, where one image per patient per magnification is extracted for computational practicality.
