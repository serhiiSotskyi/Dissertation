# BreaKHis Patient-Level Dataset

## Objective
Build a defendable patient-level split for the BreaKHis binary task and prove that the new split has no patient leakage.

## Method
Patient identifiers are parsed from the original filenames. The split operates at patient level rather than image level, because image-level random splitting leaks patient information across subsets.

```python
import sys
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd().resolve().parents[0]
sys.path.append(str(PROJECT_ROOT))

from src.breakhis import build_binary_records, cohort_summary, patient_level_split, patient_overlap_report, split_summary

dataset_root = PROJECT_ROOT / 'data' / 'dataset_cancer_v1' / 'dataset_cancer_v1'
df = build_binary_records(dataset_root)
print('images:', len(df))
print('patients:', df['patient_id'].nunique())
df.head()
```

```python
summary = cohort_summary(df)
summary
```

```python
split = patient_level_split(df, random_state=42)
split_report = split_summary(split)
overlap = patient_overlap_report(split)
print(split_report.to_string(index=False))
print(overlap)
```

## Interpretation
The overlap report must be all zeros. If any overlap exists, the split is not acceptable for the dissertation.
