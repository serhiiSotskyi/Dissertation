# Synthetic Pairing Construction

## Objective
Construct exploratory synthetic multimodal pairs from independent image and tabular datasets.

## Important limitation
Pairing is artificial and based only on class labels. These pairs do not represent real patient-level multimodal observations.

```python
import sys
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd().resolve().parents[0]
sys.path.append(str(PROJECT_ROOT))

from src.fusion import load_frozen_wisconsin_dataframe
from src.pairing import build_same_label_pairs, build_random_pairs, add_consistency_flag

wisconsin = load_frozen_wisconsin_dataframe(PROJECT_ROOT / 'notebook_Wisconsin' / 'brca.csv')
image_embeddings = pd.concat([
    pd.read_csv(PROJECT_ROOT / 'outputs_v2' / 'reports' / 'breakhis_patient_train_embeddings.csv'),
    pd.read_csv(PROJECT_ROOT / 'outputs_v2' / 'reports' / 'breakhis_patient_val_embeddings.csv'),
    pd.read_csv(PROJECT_ROOT / 'outputs_v2' / 'reports' / 'breakhis_patient_test_embeddings.csv'),
], ignore_index=True)

print('Wisconsin rows:', len(wisconsin))
print('Image embedding rows:', len(image_embeddings))
print(wisconsin['label'].value_counts().to_string())
print(image_embeddings['label'].value_counts().to_string())
```

```python
same_label_example = add_consistency_flag(build_same_label_pairs(wisconsin, image_embeddings, random_state=11).paired)
random_example = add_consistency_flag(build_random_pairs(wisconsin, image_embeddings, random_state=11).paired)
print('Same-label pair match rate:', same_label_example['pair_label_match'].mean())
print('Random pair match rate:', random_example['pair_label_match'].mean())
same_label_example.head()
```

## Interpretation
The same-label synthetic pairs are intentionally label-consistent, while the random pairs act as a methodological control. This notebook should be used to explain exactly how the synthetic dataset was manufactured and why that limits the claim.
