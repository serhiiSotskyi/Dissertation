# BreaKHis Image Baseline

## Objective
Evaluate a corrected image-only baseline using patient-level splitting and a lightweight transfer-feature strategy.

## Method
A frozen ResNet18 backbone is used to extract image embeddings from sampled BreaKHis images. The embeddings are then aggregated to patient level and classified with logistic regression. This is computationally practical on the local machine and avoids the earlier leaked split.
